# 🧠 Natural Language Processing (NLP)

## 📌 What is NLP?

**NLP (Natural Language Processing)** is a branch of **Artificial Intelligence (AI)** that helps computers understand, process, analyze, and generate **human language**.

## 🎯 Main Goals of NLP

- Understand human language
- Process and analyze text
- Extract useful information
- Understand sentiment and emotions
- Translate between languages
- Generate human-like text

## 🔧 Applications of NLP

- 🤖 Chatbots
- 🌐 Machine Translation
- 😊 Sentiment Analysis
- 📧 Spam Detection
- 🔍 Search Engines
- 📝 Text Summarization
- 🎙️ Speech Recognition
- ❓ Question Answering

## 🔄 Basic NLP Pipeline

Raw Text  
↓  
Text Cleaning  
↓  
Tokenization  
↓  
Stopword Removal  
↓  
Stemming / Lemmatization  
↓  
Feature Extraction  
↓  
Machine Learning / Deep Learning  
↓  
Prediction / Output

**Input:**
> "I really love this movie!"

**NLP Task:** Sentiment Analysis

```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
df = pd.read_csv('train.txt',sep=';',header=None,names=['text','emotion'])
```

```python
df
```

```python
df.isnull().sum()
```

```python
df['emotion'].value_counts()
```

```python
unique_emotion = df['emotion'].unique()

emotion_numbers = {}
i = 0 

for emo in unique_emotion:
    emotion_numbers[emo] = i 
    i+=1 

df['emotion'] = df['emotion'].map(emotion_numbers)
```

```python
emotion_numbers
```

```python
df.head()
```

```python
df['text']  = df['text'].apply(lambda x: x.lower())
```

```python
df.sample(5)
```

```python
import string 

def remove_punc(txt): 
    return txt.translate(str.maketrans('','',string.punctuation))
```

```python
df['text'] = df['text'].apply(remove_punc)
```

```python
df.sample(5)
```

```python
def remove_numbers(txt): 
    new = ''
    for i in txt: 
        if not i.isdigit(): 
            new = new + i 
    return new 
```

```python
df['text']  = df['text'].apply(remove_numbers)
```

```python
df.head()
```

```python
def remove_emojis(txt): 
    new = ''
    for i in txt: 
        if  i.isascii(): 
            new = new + i 
    return new 
```

```python
df['text']  = df['text'].apply(remove_emojis)
```

```python
df.head()
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
stop_words = set(stopwords.words('english'))
```

```python
stop_words
```

```python
len(stop_words)
```

```python
df.loc[1]['text']
```

```python
def remove_stop_words(txt):
    words = word_tokenize(txt)
    cleaned = []
    for i in words:
        if i not in stop_words: 
            cleaned.append(i)
    return ' '.join(cleaned)
```

```python
remove_stop_words(df.loc[1]['text'])
```

```python
df['text'] = df['text'].apply(remove_stop_words)
```

```python
df
```

```python
from sklearn.feature_extraction.text import CountVectorizer

document = df['text'].tolist()

vectorizer = CountVectorizer(max_features=1000)
x = vectorizer.fit_transform(document)
```

```python
print("Vocabulary: ",vectorizer.get_feature_names_out())
print('Bow Matrix: \n',x.toarray())
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(df['text'],
                                                 df['emotion'],
                                                 test_size=0.21,
                                                 random_state=42)
```

```python
x_train
```

```python
x_test
```

```python
print("X_train:", x_train.shape)
print("X_test:", x_test.shape)
print("Y_train:", y_train.shape)
print("Y_test:", y_test.shape)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer,CountVectorizer

from sklearn.naive_bayes import MultinomialNB 
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC

from sklearn.metrics import accuracy_score
```

```python
# CountVectorizer with MultinomialNB
bow_vectorizer = CountVectorizer()

x_train_bow = bow_vectorizer.fit_transform(x_train)
x_test_bow = bow_vectorizer.transform(x_test)

nb_model = MultinomialNB() 
nb_model.fit(x_train_bow,y_train) 

pred_nb = nb_model.predict(x_test_bow) 
print(accuracy_score(y_test,pred_nb))
```

```python
# TfidfVectorizer with MultinomialNB
tfidf_vectorizer = TfidfVectorizer(ngram_range=(1,1))

x_train_tfidf = tfidf_vectorizer.fit_transform(x_train)
x_test_tfidf = tfidf_vectorizer.transform(x_test)

nb_model = MultinomialNB() 
nb_model.fit(x_train_tfidf,y_train) 

pred_nb = nb_model.predict(x_test_tfidf) 
print(accuracy_score(y_test,pred_nb))
```

```python
# CountVectorizer with LogisticRegression
bow_vectorizer = CountVectorizer()

x_train_bow = bow_vectorizer.fit_transform(x_train)
x_test_bow = bow_vectorizer.transform(x_test)

lgr_model = LogisticRegression() 
lgr_model.fit(x_train_bow,y_train) 


pred_lgr = lgr_model.predict(x_test_bow) 
print(accuracy_score(y_test,pred_lgr))
```

```python
# TfidfVectorizer with LogisticRegression
tfidf_vectorizer = TfidfVectorizer(ngram_range=(1,1))

x_train_tfidf = tfidf_vectorizer.fit_transform(x_train)
x_test_tfidf = tfidf_vectorizer.transform(x_test)

lgr_model = LogisticRegression() 
lgr_model.fit(x_train_tfidf,y_train) 

pred_lgr = lgr_model.predict(x_test_tfidf) 
print(accuracy_score(y_test,pred_lgr))
```

```python
# CountVectorizer with LinearSVC
bow_vectorizer = CountVectorizer()

x_train_bow = bow_vectorizer.fit_transform(x_train)
x_test_bow = bow_vectorizer.transform(x_test)

svc_model = LinearSVC() 
svc_model.fit(x_train_bow,y_train) 

pred_svc = svc_model.predict(x_test_bow) 
print(accuracy_score(y_test,pred_svc))
```

```python
# TfidfVectorizer with LinearSVC
tfidf_vectorizer = TfidfVectorizer(ngram_range=(1,1))

x_train_tfidf = tfidf_vectorizer.fit_transform(x_train)
x_test_tfidf = tfidf_vectorizer.transform(x_test)

svc_model = LinearSVC() 
svc_model.fit(x_train_tfidf,y_train) 

pred_svc = svc_model.predict(x_test_tfidf) 
print(accuracy_score(y_test,pred_svc))
```

```python
# Save model and vectorizer
import joblib

joblib.dump(tfidf_vectorizer,'tfidf_vectorizer.pkl')
joblib.dump(svc_model,'svm_model.pkl')

print("Model saved successfully!")
```
