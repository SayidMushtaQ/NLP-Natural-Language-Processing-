## 🧠 Bag of Words (BoW)

**Bag of Words (BoW)** is a simple technique used in **Natural Language Processing (NLP)** to convert text into numerical data.

It works by:
- Creating a vocabulary of unique words.
- Counting how many times each word appears in a sentence.
- Representing the sentence as a vector of word counts.

> **Example:**  
> "I love RUMU" → `I:1, love:1, RUMU:1`

BoW focuses on **word frequency** and does not consider the order or meaning of words.

```python
from sklearn.feature_extraction.text import CountVectorizer

documents = [
    "I love love rumu", 
    'Rumu is the best', 
    "I love Fultosi", 
    'Fultosi is great Queen'
]

vectorizer = CountVectorizer(ngram_range=(2,2))

x = vectorizer.fit_transform(documents)
```

```python
print("Vocabulary: ",vectorizer.get_feature_names_out())
print('Bow Matrix: \n',x.toarray())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

documents = [
    "I love love rumu", 
    'Rumu is the best', 
    "I love Fultosi", 
    'Fultosi is great Queen'
]

vectorizer = TfidfVectorizer()

x = vectorizer.fit_transform(documents)
```

```python
print("Vocabulary: ",vectorizer.get_feature_names_out())
print('Bow Matrix: \n',x.toarray())
```
